# Applied AI Academy — Colab problem sets

**19 notebooks. Every problem set, with worked code and verified answers.**

Each notebook is self-contained: it downloads its dataset directly from
`appliedai.center`, so there is nothing to upload and nothing to configure.
Open one, press *Runtime → Run all*, and every answer computes in front of you.

## How to use these with a class

1. **Before the session** — run the notebook once yourself so the outputs are cached
   and you know the numbers.
2. **In the room** — show the problem cell, hide the solution, give them ten minutes.
3. **Reveal** — run the solution cell live. The answer appears with real numbers.
4. **Close on the caveat.** Every dataset here has one, and naming it is the lesson.

## Structure of every notebook

1. **Brief** — what the data is and where it came from
2. **Setup** — one cell, downloads the CSV
3. **What is in the file** — columns, types, missingness
4. **Problem 1 · Easy — describe** — task, then worked solution
5. **Problem 2 · Intermediate — build** — task, then worked solution
6. **Problem 3 · Advanced — judge** — task, then worked solution
7. **Verified answer key** — the numbers the academy ships, computed from this exact file
8. **Teaching notes** — what to draw out in discussion

## A note on the answers

Every notebook in this pack was **executed** before release, so the outputs you
see are real rather than asserted. Building this pack surfaced three errors in
the academy's own answer keys (a stale median price, wrong top-brand list, and
an over-stated modelling claim) — all three were corrected in the app. If a
future run disagrees with a key, trust the data and fix the key.
